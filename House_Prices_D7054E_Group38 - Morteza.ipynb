## STEP 4) Feature Engineering 


Add a couple of meaningful engineered features:

- `HouseAge = YrSold - YearBuilt`
- `TotalSF = TotalBsmtSF + 1stFlrSF + 2ndFlrSF`

These are important as: 
- Age is often related to price (but it can be non-linear, renovations matter etc.)
- Total square footage is a strong overall measure of size




````python


def add_engineered_features(df: pd.DataFrame) -> pd.DataFrame:
    """Add engineered features. We return a copy to avoid side effects."""
    df = df.copy()

    # 5.1 HouseAge
    # N.B: We clip negative ages to 0 just in case of odd values.
    if 'YrSold' in df.columns and 'YearBuilt' in df.columns:
        df['HouseAge'] = (df['YrSold'] - df['YearBuilt']).clip(lower=0)

    # 5.2 TotalSF
    # N.B: Handle cases where there may be missing values.
    # We fill missing with 0 so sum doesn't become NaN.
    sf_cols = ['TotalBsmtSF', '1stFlrSF', '2ndFlrSF']
    if all(col in df.columns for col in sf_cols):
        df['TotalSF'] = df[sf_cols].fillna(0).sum(axis=1)

    return df


train_fe = add_engineered_features(train_df)
test_fe = add_engineered_features(test_df)

print('New columns added:', sorted(set(train_fe.columns) - set(train_df.columns)))
train_fe[['HouseAge', 'TotalSF']].head()
````

## STEP 5) Preprocessing (cleaning + encoding)

This dataset has a mix of numeric and categorical features.

We need preprocessing because:
- Many columns have missing values
- Categorical columns must be converted into numbers

We use a scikit-learn `ColumnTransformer` pipeline:

**Numeric pipeline:**
- median imputation (simple and robust)
- standard scaling (helps linear models)

**Categorical pipeline:**
- most-frequent imputation
- one-hot encoding

> Note: Doing preprocessing inside the pipeline helps avoid leakage in CV.


````python

# Split features and target
X = train_fe.drop(columns=[TARGET_COL,'SalePrice_log'])


# ----------------------------------------------------
# Choose the relevant features
# ----------------------------------------------------
selected_features = [
    "GrLivArea",
    "GarageCars",	
    "GarageArea",	
    "TotalBsmtSF",	
    "1stFlrSF",	
    "FullBath",	
    "TotRmsAbvGrd",
    "YearBuilt",	 
    "YearRemodAdd",
    "GarageYrBlt",	
    "MasVnrArea",	
    "Fireplaces",
    "HouseAge",
    "TotalSF"
]


# This step ensures ONLY these columns are passed further into the pipeline.
feature_selector = ColumnTransformer(
    transformers=[("keep", "passthrough", selected_features)],
    remainder="drop",
    verbose_feature_names_out=False,
)

# ----------------------------------------------------
# Build a preprocessor that only sees selected features
# ----------------------------------------------------
# Build numeric/categorical lists based on the selected feature subset,
# not on the full dataset.
X_selected = X[selected_features]


# We train on log target (helps with skew)
y = np.log1p(train_fe[TARGET_COL].values)

# Identify column types
numeric_cols = X.select_dtypes(include=['number']).columns.tolist()
categorical_cols = X.select_dtypes(exclude=['number']).columns.tolist()

print('Numeric columns     :', len(numeric_cols))
print('Categorical columns :', len(categorical_cols))

# Numeric processing steps
numeric_pipe = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# Categorical processing steps
categorical_pipe = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

# Combine into one preprocessing transformer
preprocessor_selected = ColumnTransformer(
    transformers=[
        ('num', numeric_pipe, numeric_cols),
        ('cat', categorical_pipe, categorical_cols)
    ],
    remainder='drop'
)
````

## STEP 6) Model training + comparison (cross-validation)

We compare models from our proposal:

- LinearRegression (baseline)
- Ridge / Lasso (regularized linear models)
- RandomForest (ensemble of decision trees)
- GradientBoosting (boosting ensemble)

We do cross-validation for the following reasons:
- We don't want to trust a single train-test split
- CV gives us a more stable estimate of performance

We compute:
- RMSE (lower is better)
- R² (closer to 1 is better)

All evaluation is on the **log target**.


````python


def rmse(y_true: np.ndarray, y_pred: np.ndarray) -> float:
    """Compute RMSE (root mean squared error)."""
    return float(np.sqrt(mean_squared_error(y_true, y_pred)))


models = {
    'LinearRegression': LinearRegression(),

    # Student note: Ridge/Lasso have alpha which controls regularization strength.
    'Ridge': Ridge(alpha=50.0, random_state=RANDOM_STATE),
    'Lasso': Lasso(alpha=0.005, random_state=RANDOM_STATE, max_iter=5000),


    # Student note: RandomForest and GradientBoosting can capture non-linear patterns.
    'RandomForest': RandomForestRegressor(
    n_estimators=300,
    max_depth=10,              # limit tree depth
    min_samples_leaf=5,        # avoid tiny leaves
    min_samples_split=10,
    max_features='sqrt',       # classic RF setting
    random_state=RANDOM_STATE,
    n_jobs=-1
    ),
    'GradientBoosting': GradientBoostingRegressor(
    n_estimators=300,
    learning_rate=0.05,   # slower, more stable learning
    max_depth=3,          # shallow trees
    subsample=0.8,        # stochastic boosting
    random_state=RANDOM_STATE
    )
}

# CV setup
cv = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

results = []

for name, model in models.items():
    # Build full pipeline = preprocessing + model
    pipe = Pipeline(steps=[
        ("select_features", feature_selector),
        ('preprocess', preprocessor_selected),
        ('model', model)
    ])

    # Out-of-fold predictions (each sample predicted by a model that didn't train on it)
    y_pred = cross_val_predict(pipe,X, y, cv=cv, n_jobs=-1)

    results.append({
        'model': name,
        'rmse_log': rmse(y, y_pred),
        'r2_log': float(r2_score(y, y_pred))
    })

results_df = pd.DataFrame(results).sort_values(by='rmse_log')
results_df
````

## STEP 7) Best model diagnostics (plots)

After comparing models, we pick the best one (lowest RMSE).
Then we create two diagnostic plots:

1. Predicted vs Actual (should be close to a diagonal line)
2. Residuals vs Predicted (we want residuals scattered around 0)




````python

# Pick best model based on RMSE
#best_model_name = results_df.iloc[0]['model']
best_model_name = 'RandomForest'
print('Best model based on RMSE:', best_model_name)

best_pipe = Pipeline(steps=[
    ('preprocess', preprocessor),
    ('model', models[best_model_name])
])

# Out-of-fold predictions for best model
best_pred = cross_val_predict(best_pipe, X, y, cv=cv, n_jobs=-1)

# Predicted vs Actual
plt.figure(figsize=(6, 6))
plt.scatter(y, best_pred, alpha=0.4, color='teal')
min_v = min(y.min(), best_pred.min())
max_v = max(y.max(), best_pred.max())
plt.plot([min_v, max_v], [min_v, max_v], 'k--')
plt.title(f'Predicted vs Actual (log target) - {best_model_name}')
plt.xlabel('Actual (log1p)')
plt.ylabel('Predicted (log1p)')
plt.tight_layout()
plt.savefig(IMAGES_DIR / 'predicted_vs_actual.png', dpi=160)
plt.show()

# Residuals plot
residuals = y - best_pred
plt.figure(figsize=(7, 4))
plt.scatter(best_pred, residuals, alpha=0.4, color='purple')
plt.axhline(0, color='black', linestyle='--')
plt.title(f'Residuals vs Predicted (log target) - {best_model_name}')
plt.xlabel('Predicted (log1p)')
plt.ylabel('Residual (Actual - Predicted)')
plt.tight_layout()
plt.savefig(IMAGES_DIR / 'residuals.png', dpi=160)
plt.show()

print('Saved plots in:', IMAGES_DIR)
````

## STEP 8) Train final model and predict test 

Steps:
1. Fit the best pipeline on **all training data**
2. Predict on test data
3. Convert predictions back to normal price scale (`expm1`)
4. Save `prediction.csv`

N.B: Output prediction file has `Id` and `SalePrice` columns.


````python

# Fit best model on full training data
best_pipe.fit(X, y)

# Prepare test features
X_test = test_fe.copy()

# Predict log-prices then convert back to price scale
pred_test_log = best_pipe.predict(X_test)
pred_test_price = np.expm1(pred_test_log)

submission = pd.DataFrame({
    'Id': test_df['Id'],
    'SalePrice': pred_test_price
})

out_path = ROOT_DIR / 'prediction.csv'
submission.to_csv(out_path, index=False)

print('Saved:', out_path)
submission.head()

# Rename the predicted SalePrice column for clarity before merging
df_pred = submission.rename(columns={"SalePrice": "predicted_price"})

# Rename the actual SalePrice column for clarity before merging
df_actual = sample_submission_df.rename(columns={"SalePrice": "actual_price"})

# Merge actual and predicted prices on the house ID to compare them side by side
df_compare = df_actual.merge(df_pred, on='Id')

df_compare.head()

````

````python
# Calculate the raw error (positive = overestimate, negative = underestimate)
df_compare["error"] = df_compare["predicted_price"] - df_compare["actual_price"]

# Calculate the absolute error to measure magnitude regardless of direction
df_compare["abs_error"] = df_compare["error"].abs()

# Calculate the percentage error relative to the actual price
df_compare["error_pct"] = df_compare["error"] / df_compare["actual_price"] * 100

# Calculate the absolute percentage error to measure magnitude regardless of direction
df_compare["abs_error_pct"] = df_compare["error_pct"].abs()

# Preview the first few rows 
df_compare.head()
````

````python
# Calculate Mean Absolute Error (average absolute difference between predicted and actual prices)
MAE = df_compare["abs_error"].mean()

# Calculate Mean Absolute Percentage Error (average absolute % difference relative to actual prices)
MAPE = df_compare["abs_error_pct"].mean()

# Print the MAE in the same unit as the price (e.g. dollars)
print("MAE:", MAE)

# Print the MAPE as a percentage (e.g. 10.5 means predictions are off by ~10.5% on average)
print("MAPE:", MAPE)
````

````python
# Calculate the % of predictions within 0–5% error (excellent predictions)
range_0_5 = ((df_compare["abs_error_pct"] >= 0) & (df_compare["abs_error_pct"] <= 5)).mean() * 100

# Calculate the % of predictions within 6–10% error (good predictions)
range_6_10 = ((df_compare["abs_error_pct"] > 5) & (df_compare["abs_error_pct"] <= 10)).mean() * 100

# Calculate the % of predictions within 11–20% error (acceptable predictions)
range_11_20 = ((df_compare["abs_error_pct"] > 10) & (df_compare["abs_error_pct"] <= 20)).mean() * 100

# Calculate the % of predictions with >20% error (poor predictions)
range_20_plus = (df_compare["abs_error_pct"] > 20).mean() * 100

# Print the error distribution across all ranges 
print(f"0–5%: {range_0_5:.2f}%")
print(f"6–10%: {range_6_10:.2f}%")
print(f"11–20%: {range_11_20:.2f}%")
print(f">20%: {range_20_plus:.2f}%")
````

````python
# 9.1 Distribution of prediction errors 
plt.figure(figsize=(6,6))
plt.hist(df_compare["error_pct"], bins=50)
plt.xlabel("Prediction Error (%)")
plt.ylabel("Count")
plt.title("Distribution of Prediction Errors")
plt.tight_layout()
plt.savefig(IMAGES_DIR / 'prediction_errors.png', dpi=160)
plt.show()


#9.2 Predicted vs Actual scatter plot
plt.figure(figsize=(6,6))
plt.scatter(df_compare["actual_price"], df_compare["predicted_price"], alpha=0.5)
max_val = max(df_compare["actual_price"].max(), df_compare["predicted_price"].max())
plt.plot([0, max_val], [0, max_val])
plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Predicted vs Actual")
plt.tight_layout()
plt.savefig(IMAGES_DIR / 'predicted_vs_actual.png', dpi=160)
plt.show()

#9.3 Error vs Actual Price scatter plot
plt.figure(figsize=(6,6))
plt.scatter(df_compare["actual_price"], df_compare["error_pct"], alpha=0.5)
plt.axhline(0)
plt.xlabel("Actual Price")
plt.ylabel("Error (%)")
plt.title("Prediction Error vs Price")
plt.tight_layout()
plt.savefig(IMAGES_DIR / 'error_vs_actual.png', dpi=160)
plt.show()
````

## Ethics notes (short reflection)

Dataset & PII
The Ames Housing dataset has no personal information — no names, addresses, or personal IDs. It is a clean, anonymized academic dataset.

The Swedish Context
This project uses U.S. data, but it is interesting to think about what a similar model would mean in Sweden:

Offentlighetsprincipen means that house sale prices are public — anyone can look them up.
The population register links people to their addresses, so it is possible to connect a sale price to a real person.
Income data is also semi-public — you can ask Skatteverket for someone's declared income, or find it through credit score websites.
This means that in Sweden, even a simple house price model could be used to build a detailed financial profile of a real person.

Ethical Risks
Re-identification: Combining housing data with address and income registers can reveal who bought what and for how much.
Proxy discrimination: Location can reflect socio-economic status or ethnicity, which may introduce bias into predictions.
Misuse: The model could be used for purposes it was never designed for, such as credit scoring or targeted advertising.
Responsible Use
If deployed in a real Swedish context, the model should:

Follow GDPR rules on data purpose and minimisation.
Be checked for bias across different groups and areas.
Be transparent — users should know what data is used and how.
